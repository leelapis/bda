## Environment Configuration

### Installing Miniconda

Miniconda is a free, minimal installer for **conda**. It ships only conda, Python, and a few essential packages. Use it to install Python and to manage independent environments for each project or course.

- Official installer & downloads: [https://docs.conda.io/en/latest/miniconda.html](https://docs.conda.io/en/latest/miniconda.html)

- Windows: download the `Miniconda3-latest-Windows-x86_64.exe` installer and run it. Keep the default install location `C:\Users\<you>\miniconda3`.

<p class="alert alert-warning">Check the box <b>"Add Miniconda3 to my PATH environment variable"</b> during Windows installation so that <code>conda</code> works directly from the command prompt.</p>

- macOS: download the `Miniconda3-latest-MacOSX-x86_64.pkg` (Apple Silicon: `Miniconda3-latest-MacOSX-arm64.pkg`) or install via the terminal:
```bash
brew install --cask miniconda
```

- Linux: download the `.sh` installer and run `bash Miniconda3-latest-Linux-x86_64.sh` in the terminal.

#### Verifying the installation

- Open a **new** terminal window, then type:
```bash
conda --version
```
- If you see a version number (for example `conda 23.x.x`), the installation succeeded.

- Windows: press `Win + R`, type `cmd`, and press Enter to open the command prompt.
- Elegant terminal alternative: install [Windows Terminal](https://aka.ms/terminal) or use the **Anaconda Prompt** that ships with Miniconda.

### Managing Virtual Environments

A **virtual environment** is an isolated folder that owns its own Python version and its own installed packages. Projects stay independent: upgrading a package for one course never breaks another one, and each environment can be reproduced on any computer.

#### Creating & activating an environment

- Create an environment with a chosen Python version:
```bash
conda create -n bda python=3.11
```

- Activate it (prefix appears in the prompt):
```bash
conda activate bda
```
- Return to the base environment:
```bash
conda deactivate
```

#### Listing, cloning, and removing

- List all environments (the `*` marks the active one):
```bash
conda env list
```
- Remove an environment and everything inside it:
```bash
conda remove -n bda --all
```

- Install packages **inside** the active environment with `conda` or `pip`:
```bash
conda install numpy pandas matplotlib
pip install scikit-learn
```

### Package Management Tools (pip)

<p class="alert alert-info">pip is the package installer for Python. It finds, downloads, installs, and uninstalls Python packages.</p>

- On `macOS`/`Linux` use `pip3` (or simply `pip` inside a conda environment).

- Install the latest version of a package: `pip install -U <package_name>`

- Uninstall a package: `pip uninstall <package_name>`

- List installed packages: `pip list`

#### Using packages

##### Importing a complete package

- Use the `import` statement before your code
```python
import pandas
```

- Use the imported package
```python
pandas.DataFrame(X)
```

- Give the imported package a short alias
```python
import pandas as pd
```

- Use the aliased package
```python
pd.DataFrame(X)
```

- Import several packages, one statement per line
```python
import pandas as pd
import numpy as np
```

##### Importing a specific module of a package

- Use the `from ... import ...` statement
```python
from datetime import datetime
```

- Use the imported module
```python
datetime.now()
```

## Markdown

- A lightweight **markup language**

> We believe that writing is about content, about what you want to say – not about fancy formatting.

### Advantages of Markdown

- Focus on the content of the text rather than its layout

- Exports to many file types, including HTML, PDF, Word, and LaTeX

- Simple rules, few mark symbols

### Basic syntax

#### Headings

- Prefix a line with `#`. The number of `#` signs sets the heading level, with a space between the last `#` and the text
- Markdown supports 6 heading levels

```markdown
# Heading 1

## Heading 2

### Heading 3
```

#### Lists

- Unordered lists: prefix items with `-` or `*` (a space must follow the symbol)

- Ordered lists: prefix items with `1. 2. 3.` (a space must follow the period)

```
- Point 1
- Point 2
- Point 3
```

```
1. Classification
2. Clustering
3. Association
```

- Nested lists: indent the sub-list with one Tab key before it

1. Classification
    1. Decision tree
    2. Neural network
2. Clustering
    1. k-means
    2. Hierarchical

#### Bold, italic & strikethrough

- Wrap text in <span class="text text-danger">two</span> `*` to make it <span class="mark">bold</span>

- Wrap text in <span class="text text-danger">one</span> `*` to make it <span class="mark">italic</span>

- Wrap text in <span class="text text-danger">two</span> `~~` to draw a strikethrough over it

```
*Python* is **very important** for employment, ~~drop this sentence~~
```

*Python* is **very important** for employment, ~~drop this sentence~~

#### Line breaks

- Separate paragraphs with a blank line

```
This is the first paragraph.

This is the second paragraph.
```

This is the first paragraph.

This is the second paragraph.

#### Blockquotes

- Prefix a line with `>` to turn it into a blockquote

- Use multiple `>` for nested blockquotes

```
This is body text
> This is quoted content
>> This is a nested quote
```

This is body text
> This is quoted content
>> This is a nested quote

#### Images

- Insert an image: `![alternative text](image path or URL)`

```
![](https://upload.wikimedia.org/wikipedia/commons/7/75/Kawaii_robot_power_clipart.svg)
```

![](https://upload.wikimedia.org/wikipedia/commons/7/75/Kawaii_robot_power_clipart.svg)

#### Links

- Insert a link: `[link text](URL)`

```
[Python](https://www.python.org/)
```

[Python](https://www.python.org/)

#### Tables

- Separate columns with `|`

- Separate the header from the body with `-`; each column needs at least three `-`

```
ID|Employment|Education|Years of residence|Credit status
---|---|---|---|---
1|Employed|Master's degree|5|Good
2|Employed|High school|2|Poor
```
