##### AI-Era Scams: When the Data Itself Is Fake

AI turned online fraud into an **industry** — fake texts, photos, voices, invoices
and “job offers” can be generated at scale. For an analyst, this is an attack on
**Veracity**: the data may look perfect, but the world that produced it is now full
of fabrications.

**Headline damage (US, 2025 — FBI IC3):**
- **$20.9 billion** lost to online scams in a single year — a **5× jump** since 2020.
- **22,000+** complaints were linked to suspected AI involvement, totalling **$893 million** in losses.
- The most profitable AI-enhanced fraud was **investment scams**: **$8.6B** lost, **$7.2B** of it in crypto.


##### Reported losses by scam family (2025)

| Scam family | How AI amplifies it | Reported 2025 damage |
|---|---|---|
| Romance / sextortion | Fake personas, synthetic chats; top complaint for minors | ~**$1B** |
| Investment & crypto | Fake “winners”, lifelike apps & charts | **$8.6B** (crypto $7.2B) |
| Government impersonation | Caller-ID spoofing + cloned voices/video | ~**$800M** (2× 2024) |
| Employment | Fake recruiters, “laptop farming”, fake remote jobs | **$363M** |
| Invitation phishing | Hacked accounts send invites that look real | part of the $20.9B total |

> Figures: U.S. FBI IC3 annual report, cited in *Bloomberg Businessweek*, “How to Not Get Scammed in the Age of AI” (August 2026).


**What this teaches about Veracity:**

- In an AI world, “trust but verify” is dead — **“verify, verify again, then maybe trust”** (Nils Mueller, Global Anti-Scam Alliance).
- The more “professional” and “perfect” the data looks, the harder you should look for its **source and provenance**.
- Practical analyst habits: check the sender address before clicking links, confirm identity on a video call before paying, agree a **safe word** with family, and **search “name + scam/review/complaint”** before trusting an offer.

### Types of Attributes

- The type of an attribute depends on which of the following properties its values exhibit:   
   + **Distinctness**: $=$ and $\ne$    
   + **Order**: $\lt$, $\le$, $\gt$, and $\ge$    
   + **Meaningful differences**: $+$ and $-$    
   + **Meaningful ratios**: $\times$ and $/$

|Attribute Type|Characteristics|Properties|Example|Appropriate Statistics|
|---|---|---|---|---|
|Nominal|Mere names|Distinctness|Name, class, color|Mode|
|Ordinal|Natural ordering|Distinctness, order|Grade, rank|Mode, median|
|Interval|No absolute zero, equal intervals|Distinctness, order, meaningful differences|Temperature, date, IQ|Mode, median, mean|
|varianceRatio|Absolute zero|Distinctness, order, meaningful differences, meaningful ratios|Distance, weight, income|Mode, median, mean, variance, geometric mean|

#### Phase 1-2: Business Understanding & Data Understanding

**Business Understanding** — the most critical phase
- Translate business problems into data questions
- Define success metrics (KPIs, OKRs)
- Assess feasibility: do we have the right data?
- AI feasibility check: is LLM/ML the right solution?

**Data Understanding**
- Inventory available data sources
- Profile data: types, distributions, missing values
- Assess data quality: completeness, consistency, accuracy
- Modern tools: data catalogs (Alation, Atlan), data lineage (dbt, Monte Carlo)

> <span class="girk">Without business understanding, you build models that answer the wrong question</span>

##### Data Collection — Primary & Secondary Sources

**Primary Data** (collected first-hand)
- Surveys & interviews
- Experiments & A/B tests
- Web scraping & API harvesting
- IoT sensors & logs
- AI-generated synthetic data

**Secondary Data** (existing sources)
- Financial databases: Wind, CSMAR, Choice
- Open data portals: government, WHO, World Bank
- APIs: Tushare, AKShare, social media APIs
- Third-party data providers

#### Phase 3: Data Preparation (ETL → ELT → dbt)

**Traditional approach:** Transform raw data into structured format<br>
| id | Name | Gender | Age | Student ID | Score |
|---|---|---|---|---|---|
| 1 | Tom | M | 21 | 20170201 | 90 |

**Modern Data Preparation (2025-2026):**
- ELT (Extract-Load-Transform) instead of ETL — load raw data first, transform later
- dbt (data build tool): transform data in warehouse using SQL
- Data quality checks: Great Expectations, dbt tests
- Feature stores (Feast, Tecton): reusable ML features
- DataOps: CI/CD for data pipelines, automated testing

**Data Cleaning** — still the most time-consuming step (~80% of project effort)
- Outlier detection & treatment
- Invalid data removal
- Duplicate deduplication
- Missing value imputation
- Modern: automated profiling (Pandas Profiling, ydata-profiling)

#### Phase 4: Exploratory Data Analysis (EDA)

**Descriptive Statistics**
- Mean, median, standard deviation, quantiles
- Distribution analysis, skewness, kurtosis
- Correlation analysis

**Data Visualization**
- Histograms → distribution
- Scatter plots → relationships
- Line charts → trends over time
- Heatmaps → correlation matrices
- Modern: interactive dashboards (Plotly, Streamlit)

**AI-Augmented EDA (2026)**
- LLM-assisted data profiling: auto-generate insights from data
- Agentic analytics: AI agent plans, queries, and explains EDA results
- Auto-visualization: AI recommends chart types based on data patterns
- Tools: DA-Studio, LIDA, pandas-ai

#### Phase 5: Modeling & Algorithms

**Statistical & Econometric Models**
- Linear / Logistic regression
- Factor analysis, PCA
- Time series analysis (ARIMA, Prophet)

**Machine Learning**
- Supervised learning (classification, regression)
- Unsupervised learning (clustering, dimensionality reduction)
- Reinforcement learning

**Modern ML Practices (2025-2026):**
- AutoML: automated model selection & hyperparameter tuning
- LLMs & Generative AI: RAG, fine-tuning, prompt engineering
- Experiment tracking: MLflow, Weights & Biases
- MLOps: version control for models, reproducible pipelines
- Choose based on problem type: classification, prediction, or explanation?

#### Phase 6: Evaluation & Deployment

**Evaluation**
- Technical: accuracy, precision, recall, F1, AUC-ROC
- Business: did we meet the success metrics?
- Fairness audits: bias detection & mitigation
- Explainability: SHAP, LIME, feature importance

**Deployment**
- Model serving: REST APIs, batch inference, edge deployment
- CI/CD pipelines: automated testing & deployment
- Model monitoring: drift detection, performance decay
- CRISP-ML(Q): quality assurance throughout the lifecycle

#### Ethics, Governance & Responsible AI

## Course Structure


###  A business-first storyline

- This course starts from a common analytics question: how do we turn messy operational data into decisions that improve revenue, service, and customer experience?

- We use hotel booking problems as the running example because they contain the full workflow: data cleaning, demand patterns, guest segmentation, cancellation risk, and pricing.

<img src="./img/introduction/course_structure_final.png" width="100%">


**1. Build the data foundation**

- Advanced Pandas and Polars to prepare reliable business data
- Data quality checks: missing values, outliers, and automated EDA
- GenAI-assisted coding to speed up routine cleaning
- Hotel booking data is the main practice case for these skills

> Business question: Can we trust our raw booking records?


**2. Discover patterns and guest segments**

- High-dimensional pattern discovery (PCA, t-SNE, UMAP)
- Customer segmentation (K-Means, DBSCAN) to find hidden guest groups
- Interpreting why segments are different using statistical summaries

> Business question: Are there hidden dimensions? Who are our natural segments?


**3. Predict outcomes and explain value**

- Supervised learning: predicting cancellations (Classification)
- Forecasting demand and revenue (Regression)
- Explainable AI: using SHAP values to tell stakeholders *why* a guest might cancel

> Business question: Will this specific guest cancel? What price should we set?


**4. Deliver decisions that stakeholders can use**

- The industry project combines cleaning, exploration, modeling, and deployment
- Teams define KPIs, build a pipeline, and present the results in a Streamlit dashboard
- The final output is a business recommendation, not just a model score

> Business question: How do we optimize hotel profit using these insights?
